```python
excel_file_name = 'IPL sample data.xlsx'

try:
    # Read the Excel file into a pandas DataFrame, skipping initial rows.
    # The true headers appear to be on the 4th row (index 3) of the Excel file.
    # We'll then manually set these as columns.
    df_excel = pd.read_excel(excel_file_name, header=3)

    # The actual column names are in the first row of the loaded DataFrame (df_excel.iloc[0])
    df_excel.columns = df_excel.iloc[0]
    # Drop the row that was just used as headers
    df_excel = df_excel[1:]
    # Reset the index
    df_excel = df_excel.reset_index(drop=True)

    print(f"Successfully loaded '{excel_file_name}' into a DataFrame with correct headers.")
    display(df_excel.head())
except FileNotFoundError:
    print(f"Error: The file '{excel_file_name}' was not found. Please ensure it's uploaded and the name is correct.")
except Exception as e:
    print(f"An error occurred while reading the Excel file: {e}")
```

```python
print("Current columns in df_excel after header fix:")
print(df_excel.columns.tolist())
```

```python
# Drop columns where all values are NaN
df_excel = df_excel.dropna(axis=1, how='all')

print("DataFrame after dropping all-NaN columns:")
display(df_excel.head())

print("Current columns in df_excel after dropping all-NaN columns:")
print(df_excel.columns.tolist())
```

```python
display(df_excel.describe(include='all'))
```

```python
df_filtered = df_excel[df_excel['Player Name'].notna()]
print("DataFrame after filtering for 'Player Name' not null:")
display(df_filtered.head())
```
